1. A random sample has

   $$
   \bar{X}=52,\qquad s=10,\qquad n=100.
   $$

   Compute the estimated standard error of $\bar{X}$. Test

   $$
   H_0:\mu=50
   \qquad \text{vs.} \qquad
   H_1:\mu\ne 50
   $$

  ANSWER:

The estimated standard error is:

$$
\widehat{SE}(\bar X)=\frac{s}{\sqrt{n}}
=\frac{10}{\sqrt{100}}=1.
$$

The test statistic is:

$$
z=\frac{\bar X-\mu_0}{\widehat{SE}}
=\frac{52-50}{1}=2.
$$

For the two-sided test, compare $|z|=2$ with each critical value:

- At $\alpha=0.10$: $2>1.645$, so reject $H_0$.
- At $\alpha=0.05$: $2>1.96$, so reject $H_0$.
- At $\alpha=0.01$: $2<2.576$, so fail to reject $H_0$.

The 95% confidence interval for the population mean is:

$$
\bar X \pm 1.96\widehat{SE}
=52\pm1.96(1)
=[50.04,\ 53.96].
$$

The interval excludes 50, consistent with rejecting the null hypothesis
at the 5% significance level.

2. A health survey compares a biomarker to a benchmark value of 12. The sample has

   $$
   \bar{X}=11.5,\qquad s=3,\qquad n=36.
   $$

   ANSWER:

The estimated standard error is:

$$
\widehat{SE}=\frac{s}{\sqrt{n}}
=\frac{3}{\sqrt{36}}=0.5.
$$

The 95% confidence interval is:

$$
\bar X\pm1.96\widehat{SE}
=11.5\pm1.96(0.5)
=[10.52,\ 12.48].
$$

For the two-sided test of $H_0:\mu=12$ versus $H_1:\mu\ne12$:

$$
z=\frac{11.5-12}{0.5}=-1.
$$

The two-sided p-value is:

$$
p=2\Phi(-|z|)=2\Phi(-1)\approx0.3173.
$$

Since $0.3173>0.05$, we fail to reject the null hypothesis.
The value 12 is also inside the 95% confidence interval, so both
the interval and the test indicate that 12 is a plausible population
mean. This does not prove that the population mean equals 12.

3. A real estate analyst compares the average price of a house in two neighborhoods.
   Group A:

   $$
   \bar{X}_A=25,\qquad s_A=8,\qquad n_A=100
   $$

   Group B:

   $$
   \bar{X}_B=22,\qquad s_B=7,\qquad n_B=100.
   $$

   Compute the two-sample $Z$-statistic for testing equality of means. Construct an approximate 95% confidence interval for $\mu_A-\mu_B$.

   ANSWER:

The estimated difference between the population means is:

$$
\bar X_A-\bar X_B=25-22=3.
$$

Assuming independent samples, the estimated standard error is:

$$
\widehat{SE}
=\sqrt{\frac{s_A^2}{n_A}+\frac{s_B^2}{n_B}}
=\sqrt{\frac{8^2}{100}+\frac{7^2}{100}}
=\sqrt{1.13}
\approx1.0630.
$$

For testing $H_0:\mu_A-\mu_B=0$ against
$H_1:\mu_A-\mu_B\ne0$, the test statistic is:

$$
Z=\frac{(25-22)-0}{\sqrt{1.13}}
\approx2.8222.
$$

The approximate 95% confidence interval is:

$$
(\bar X_A-\bar X_B)\pm1.96\widehat{SE}
=3\pm1.96\sqrt{1.13}
\approx[0.9165,\ 5.0835].
$$

Calculations use the unrounded standard error.

The interval excludes zero, so we reject equality of the population
means at the 5% significance level. The estimated average house price
in neighborhood A is 3 price units higher than in neighborhood B.

4. For `./data/ames_prices.csv`, define an indicator equal to 1 when `Central.Air == 'Y'`. Estimate the proportion of homes with central air and test $H_0:p=.90$. Treat a proportion as the mean of a binary variable and complete the full inference routine.

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

ames = pd.read_csv("ames_prices.csv")

central_air = (ames["Central.Air"] == "Y").astype(int).to_numpy()
p_hat = central_air.mean()

seed = 5030
R = 2000
rng = np.random.default_rng(seed)

bootstrap_proportions = np.array([
    rng.choice(central_air, size=len(central_air), replace=True).mean()
    for _ in range(R)
])

se = bootstrap_proportions.std(ddof=1)
z = (p_hat - 0.90) / se
p_value = 2 * norm.cdf(-abs(z))

print(f"Seed: {seed}; R: {R}")
print(f"Estimated proportion: {p_hat:.6f}")
print(f"Bootstrap standard error: {se:.6f}")
print(f"z-statistic: {z:.4f}")
print(f"Two-sided p-value: {p_value:.6g}")

for confidence, alpha, critical in [
    (90, 0.10, 1.645),
    (95, 0.05, 1.96),
    (99, 0.01, 2.576)
]:
    lower = p_hat - critical * se
    upper = p_hat + critical * se
    contains_null = lower <= 0.90 <= upper
    reject = p_value < alpha

    print(f"\n{confidence}% CI: [{lower:.6f}, {upper:.6f}]")
    print(f"At alpha = {alpha}: "
          f"{'Reject H0' if reject else 'Fail to reject H0'}")
    print(f"CI contains 0.90: {contains_null}")
    print(f"Test and CI agree: {reject == (not contains_null)}")
```

ANSWER: 

Using 2,000 bootstrap resamples with seed 5030, the estimated proportion
of homes with central air is approximately 0.9331, or 93.31%.
The bootstrap standard error is approximately 0.0046.

The centered-Normal confidence intervals are approximately:

- 90%: [0.9255, 0.9407]
- 95%: [0.9241, 0.9422]
- 99%: [0.9212, 0.9450]

These intervals give plausible values for the population proportion
under the sampling assumptions.

The two-sided test of $H_0:p=0.90$ gives a z-statistic of approximately
7.1682 and a p-value below 0.0001. We reject the null hypothesis at
significance levels 0.10, 0.05, and 0.01.

All three confidence intervals exclude 0.90, so each interval agrees
with its corresponding hypothesis-test decision.

The evidence indicates that the population proportion of homes with
central air differs from 90%, with the estimate about 3.31 percentage
points higher. This describes central-air prevalence and does not
establish causation.

5. For `./data/ames_prices.csv`, estimate the Pearson correlation between `arcsinh(area)` and `arcsinh(price)`. Bootstrap paired rows, construct all three CIs, and test $H_0:\rho=0$. Explain what would go wrong if area and price were resampled separately.

```python
# Keep area and price paired for each home.
paired = ames[["area", "price"]].dropna()
transformed = np.arcsinh(paired.to_numpy())

r_hat = np.corrcoef(transformed[:, 0], transformed[:, 1])[0, 1]

seed = 5030
R = 2000
rng = np.random.default_rng(seed)

bootstrap_correlations = np.empty(R)

for i in range(R):
    indices = rng.choice(len(transformed), size=len(transformed), replace=True)
    sample = transformed[indices]
    bootstrap_correlations[i] = np.corrcoef(sample[:, 0], sample[:, 1])[0, 1]

se5 = bootstrap_correlations.std(ddof=1)
z5 = r_hat / se5
p_value5 = 2 * norm.cdf(-abs(z5))

print(f"Seed: {seed}; R: {R}")
print(f"Estimated correlation: {r_hat:.6f}")
print(f"Bootstrap standard error: {se5:.6f}")
print(f"z-statistic: {z5:.4f}")

if p_value5 == 0:
    print("Two-sided p-value: too small to represent; not mathematically zero.")
else:
    print(f"Two-sided p-value: {p_value5:.6g}")

for confidence, alpha, critical in [
    (90, 0.10, 1.645),
    (95, 0.05, 1.96),
    (99, 0.01, 2.576)
]:
    lower = r_hat - critical * se5
    upper = r_hat + critical * se5
    contains_null = lower <= 0 <= upper
    reject = p_value5 < alpha

    print(f"\n{confidence}% CI: [{lower:.6f}, {upper:.6f}]")
    print(f"At alpha = {alpha}: "
          f"{'Reject H0' if reject else 'Fail to reject H0'}")
    print(f"CI contains zero: {contains_null}")
    print(f"Test and CI agree: {reject == (not contains_null)}")
```

ANSWER:

The estimated Pearson correlation between arcsinh(area) and
arcsinh(price) is approximately 0.7232, indicating a fairly strong
positive linear association between the transformed variables.

Using 2,000 paired-row bootstrap resamples with seed 5030, the
bootstrap standard error is approximately 0.0100.

The centered-Normal confidence intervals are approximately:

- 90%: [0.7067, 0.7396]
- 95%: [0.7035, 0.7428]
- 99%: [0.6974, 0.7490]

All three intervals exclude zero. The two-sided p-value is extremely
small, so we reject $H_0:\rho=0$ at significance levels 0.10, 0.05,
and 0.01. Each test decision agrees with its corresponding interval.
A displayed p-value of zero reflects numerical underflow, not an
exactly zero probability.

Resampling paired rows preserves the relationship between each home's
area and price. Resampling area and price separately would destroy
those pairings and generally produce correlations centered near zero,
rather than estimate uncertainty around the observed correlation.

Larger homes tend to have higher prices, but this observational
association does not establish causation. Other factors, such as
location, could contribute to the relationship.

standard error is approximately 2.2045 months.

The centered-Normal confidence intervals are approximately:

| Confidence level | Interval in months | Interval in years |
|---|---|---|
| 90% | [125.4679, 132.7207] | [10.4557, 11.0601] |
| 95% | [124.7735, 133.4151] | [10.3978, 11.1179] |
| 99% | [123.4155, 134.7731] | [10.2846, 11.2311] |

These intervals give plausible values for the population mean
recorded time to death or last follow-up under the sampling assumptions.

The two-sided test of $H_0:\mu=120$ gives a z-statistic of approximately
4.1253 and a p-value below 0.0001. We reject the null hypothesis at

All three intervals exclude 120 months (10 years), so every interval

```python
nhanes = pd.read_csv("nhanes_data_17_18.csv", low_memory=False)

# Adults aged 18 or older with observed weekday/workday sleep.
sleep = nhanes.loc[
    (nhanes["AgeInYearsAtScreening"] >= 18)
    & nhanes["SleepHoursWeekdaysOrWorkdays"].notna(),
    "SleepHoursWeekdaysOrWorkdays"
].to_numpy()

mean7 = sleep.mean()

seed = 5030
R = 2000
rng = np.random.default_rng(seed)

bootstrap_means7 = np.array([
    rng.choice(sleep, size=len(sleep), replace=True).mean()
    for _ in range(R)
])

se7 = bootstrap_means7.std(ddof=1)
z7 = (mean7 - 8) / se7
p_value7 = 2 * norm.cdf(-abs(z7))

print(f"Seed: {seed}; R: {R}")
print(f"Number of adults with observed sleep: {len(sleep)}")
print(f"Estimated mean sleep: {mean7:.6f} hours")
print(f"Difference from 8 hours: {(mean7 - 8) * 60:.2f} minutes")
print(f"Bootstrap standard error: {se7:.6f} hours")
print(f"z-statistic: {z7:.4f}")
print(f"Two-sided p-value: {p_value7:.6g}")

for confidence, alpha, critical in [
    (90, 0.10, 1.645),
    (95, 0.05, 1.96),
    (99, 0.01, 2.576)
]:
    lower = mean7 - critical * se7
    upper = mean7 + critical * se7
    contains_null = lower <= 8 <= upper
    reject = p_value7 < alpha

    print(f"\n{confidence}% CI in hours: [{lower:.6f}, {upper:.6f}]")
    print(f"At alpha = {alpha}: "
          f"{'Reject H0' if reject else 'Fail to reject H0'}")
    print(f"CI contains 8 hours: {contains_null}")
    print(f"Test and CI agree: {reject == (not contains_null)}")
```
