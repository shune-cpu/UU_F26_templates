HTML is built on TAGS

Syntax is:

'<tag attribute=value> navigable string</tag>'

It's possible for your data to live in the attribute's value, or the navigable string.

Our task: identify the tag, then extract the data we want

```python
tags = mysoup.find_all('poster-tile', attrs = {'media-url': True})
tags
```

```python
# if the data were in the navigable string you would do
# tags [0].string or tags[0].text

# Because the data is an attribute balue we do this:
tags[0] ['media-url']
```

```python
movie_urls = ['https://rottentomatoes.com' + t['media-url'] for t in tags]
```

```python
r = requests.get(movie_urls[72], headers=headers)
r
```

```python
mysoup = BeautifulSoup(r.text, 'html.parser')
mysoup
```

```python
movie_dict = {}

# Title
title = mysoup.find('title')
title.string
movie_dict['title'] = title
movie_dict
```
